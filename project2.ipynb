```python
# STEP NO 1 DATA COLLECTION AND OPENING THE DATA
with open("finaldata.txt", encoding = 'utf-8') as f:
    data = f.read()
```

```python
data
```

```python
chunks = data.split("\n\n")
chunks = [c for c in chunks if len(c) > 3]
```

```python
def parse_chunk(chunk):
        chunk = chunk.strip()
        seperate_chunk = chunk.split('\n')
        user_name = seperate_chunk[0]
        no_of_posts = int(seperate_chunk[1].split(" post")[0].replace(",", ""))
    
        no_of_followers = float(seperate_chunk[2].split("followers")[0].replace(",", "").replace("K", "") .replace("M", ""))
        if ("K" in seperate_chunk[2]):
            no_of_followers = int(no_of_followers * 1000)
        elif ("M" in seperate_chunk[2]):
            no_of_followers = int(no_of_followers * 1000000)
        else:
            no_of_followers = int(no_of_followers)
    
        no_of_following = float(seperate_chunk[3].split("following")[0].replace(",", "").replace("K", "") .replace("M", ""))
    
        if ("K" in seperate_chunk[3]):
            no_of_following = int(no_of_following * 1000)
        elif ("M" in seperate_chunk[3]):
            no_of_following = int(no_of_following * 1000000)
        else:
            no_of_following = int(no_of_following)
        name = seperate_chunk[4]
        if(len(seperate_chunk)>6):
            type_of_page = seperate_chunk[5]
            bio = "\n".join(seperate_chunk[6:])
        else:
            type_of_page = "Unknown"
            bio = "Not written"
        return {"user_name": user_name, "no_of_posts" : no_of_posts, "no_of_followers":no_of_followers, "no_of_following": no_of_following, "type_of_page":type_of_page}
```

```python
all_chunks =[]
for chunk in chunks:
    parsed_chunk = parse_chunk(chunk)
    all_chunks.append(parsed_chunk)
```

```python
import json
s =json.dumps(all_chunks, indent = 3)
with open("data.json", "w") as f:
    f.write(s)
```
